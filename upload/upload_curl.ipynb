```python
# Initialize, upload, commit, and publish files in folder 83483
import os
folder_path = "../downloads/83236"

url = "https://workflow-repo.test.du.cesnet.cz/api/records"
headers = {
    "Authorization": f"Bearer {token}",
    "Content-Type": "application/json"
}
headers_upload = {
    "Authorization": f"Bearer {token}",
    "Content-Type": "application/octet-stream"
}
with open("minimal_record.json", "r", encoding="utf-8") as f:
    data = json.load(f)
response = requests.post(url, headers=headers, json=data)
record = response.json()
record_id = record["id"]
print("Record ID:", record_id)

# add record_id to records.txt
with open("../records.txt", "a", encoding="utf-8") as f:
    f.write(record_id + "\n")

record_id = "5qpnz-1sn87"

# Initialize file upload
url = f"https://workflow-repo.test.du.cesnet.cz/api/records/{record_id}/draft/files"
data = []
for i, filename in enumerate(os.listdir(folder_path)):
    file_path = os.path.join(folder_path, filename)
    if os.path.isfile(file_path):
        data.append({"key": filename})
print(json.dumps(data, indent=2, ensure_ascii=False))
response = requests.post(url, headers=headers, json=data)
print(f"Initialized upload for folder {folder_path}:")
print(json.dumps(response.json(), indent=2, ensure_ascii=False))

for filename in os.listdir(folder_path):
    file_path = os.path.join(folder_path, filename)
    if os.path.isfile(file_path):
        # Upload the actual file
        url = f"https://workflow-repo.test.du.cesnet.cz/api/records/{record_id}/draft/files/{filename}/content"
        with open(file_path, "rb") as f:
            file_data = f.read()
        response = requests.put(url, headers=headers_upload, data=file_data)
        print(f"Uploaded {filename}:")
        print(json.dumps(response.json(), indent=2, ensure_ascii=False))
        # Commit the uploaded file
        url = f"https://workflow-repo.test.du.cesnet.cz/api/records/{record_id}/draft/files/{filename}/commit"
        response = requests.post(url, headers={"Authorization": f"Bearer {token}"})
        print(f"Committed {filename}:")
        print(json.dumps(response.json(), indent=2, ensure_ascii=False))
        # break
        # Publish the record after each file commit
url = f"https://workflow-repo.test.du.cesnet.cz/api/records/{record_id}/draft/actions/publish"
response = requests.post(url, headers={"Authorization": f"Bearer {token}"})
print(f"Published record after committing {filename}:")
print(json.dumps(response.json(), indent=2, ensure_ascii=False))
```
