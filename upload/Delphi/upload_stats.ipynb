```python
import pandas as pd

stats_path = "upload_stats.csv"

df = pd.read_csv(stats_path)

df.head(10)
```

```python
# convert start_ts to datetime
df['start_ts'] = pd.to_datetime(df['start_ts'])
```

```python
# all unique recids minus the ones with status ok

unique_recids = df['recid'].unique()
status_ok_recids = df[df['status'] == 'ok']['recid'].unique()
recids_without_ok = unique_recids[~pd.Series(unique_recids).isin(status_ok_recids)]

print(f"Number of unique recids without 'ok' status: {len(recids_without_ok)}")
```

```python
# Print all errors for each recid without 'ok' status

for recid in recids_without_ok:
    errors = df[(df['recid'] == recid) & (df['status'] != 'ok')]
    print(f"Recid: {recid}")
    print(errors['error'])
    print("\n")
```
