```python
### This code removes obsolete data from the file download_stats.csv.
### Each download appends data, so the stats reflect all runs combined. 
### To remove obsolete data, enter the date of the run you want to keep.
```

```python
import pandas as pd
import numpy as np
```

```python
stats_file = pd.read_csv('download_stats.csv')
```

```python
stats_file.tail()
```

```python
stats_file['start_date'] = pd.to_datetime(stats_file['start'], unit = 's')
stats_file['start_date'] = stats_file['start_date'].dt.floor('D')  #
```

```python
stats_file['start_date'].unique()
stats_file.groupby('start_date').count()
```

##Fill in the date of the run to be kept:

```python
stats_file = stats_file[stats_file['start_date'] == '2026-04-14']
```

```python
stats_file.groupby('start_date').count()
```

```python
stats_file.drop(columns = ['start_date'], inplace = True)
stats_file.head()
```

```python
stats_file.to_csv('download_stats.csv', index=False)
```
